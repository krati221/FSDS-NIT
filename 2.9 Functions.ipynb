```python
def add_sub(x,y):
  c = x+y
  d = x-y
  return c,d
result,result1 = add_sub(4,5)
print(result)
print(result1)
print(type(result))
print(type(result1))
```

```python
def add_sub_mul(x,y):
  c = x + y
  d = x - y
  e = x * y
  return c,d,e
result,result1,result2 = add_sub_mul(4,5)
print(result)
print(result1)
print(result2)
```
